# PhonePe Pulse — Exploratory Data Analysis

This notebook explores the processed PhonePe Pulse transaction dataset.

**Before running:** execute `python run_pipeline.py` from the project root so that the processed CSV files exist in `data/processed/`.

```python
from pathlib import Path

import pandas as pd

DATA = Path("../data/processed")
tx = pd.read_csv(DATA / "aggregated_transactions.csv")

tx.head()
```

```python
quarterly = (
    tx.groupby(["year", "quarter"], as_index=False)
    .agg(
        transactions=("transaction_count", "sum"),
        value=("transaction_amount", "sum"),
    )
)

quarterly["value_crore"] = quarterly["value"] / 1e7
quarterly.tail(10)
```

```python
category = (
    tx.groupby("category", as_index=False)
    .agg(
        transactions=("transaction_count", "sum"),
        value=("transaction_amount", "sum"),
    )
    .sort_values("transactions", ascending=False)
)

category
```
